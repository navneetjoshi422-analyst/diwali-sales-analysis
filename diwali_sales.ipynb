# diwali_sales Analysis using python


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Diwali Sales Data.csv',encoding='unicode_escape')
df.shape 
```

```python
df.head(10)
```

```python
df.describe()
```

```python
df.info()
```

```python
## data cleaning in python
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
# check null values in a data frame 
# but it retruns data frame with boolean values
df.isnull()
```

```python
# check if there is any null value present in this dataframe 
df.isnull().sum()
#pd.isnull(df).sum()
```

```python
#drop null values
df.dropna(inplace=True)
```

```python
# change the data type of a column
df["Amount"]=df["Amount"].astype('int')
```

```python
df['Amount'].dtypes
```

```python
# rename column
df.rename(columns={'Marital_Status':'shaadi'})
```

```python
df.describe()
```

```python
#describe function for specific columns
df[['Age','Orders','Amount']].describe()
```

## EDA [EXPLORATORY DATA ANALYSIS]

```python
df.columns
```

```python
ax=sns.countplot(x='Gender',data=df,palette="viridis")
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sales_gen)
```

```python
sales_gen
```

## from the above graph i can see that the most buyers are female and even the purchasing value is greater then men

## age

```python
ax=sns.countplot(data=df,x='Age Group',hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_genn_by_age_group=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sales_genn_by_age_group
```

```python
sns.barplot(x='Age Group',y='Amount',data=sales_genn_by_age_group,palette="viridis")
```

## from the above graph most of the sale is done by age group 26-35 females 

# State 

```python
sales_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(16,5)})
sns.barplot(x='State',y='Orders',data=sales_state,palette="viridis")
```

### most of the orders placed by Uttar pardesh and all the top 10 states are shown in the graph

```python
total_sales_state=df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(16,5)})
sns.barplot(x='State',y='Amount',data=total_sales_state,palette="viridis")
```

### most of the amount spend by Uttar pardesh and all the top 10 states are shown in the graph

## marital Status

```python
ax=sns.countplot(data=df,x='Marital_Status')
sns.set(rc={'figure.figsize':(7,4)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df.columns
```

```python
total_sales_MS=df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False) 
sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(x='Marital_Status',y='Amount',data=total_sales_MS,hue='Gender')
```

from above graph we can see that most of the buyers are married (females)

# Occupation

```python
sns.set(rc={'figure.figsize':(20,5)})
ax=sns.countplot(data=df,x='Occupation')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
total_sales_occupation=df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10) 
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Occupation',y='Amount',data=total_sales_occupation)
```

from above graphs we know that most of the buyers are from it healthcare and aviation sector

# Product category

```python
sns.set(rc={'figure.figsize':(25,4)})
ax=sns.countplot(data=df,x='Product_Category')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
total_sales_occupation=df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10) 
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Product_Category',y='Amount',data=total_sales_occupation)
```

```python
total_sales_occupation=df.groupby(['Product_ID'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10) 
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Product_ID',y='Orders',data=total_sales_occupation)
```

from above graphs we can see that most of the sold products are from food fpptwear and electonic
